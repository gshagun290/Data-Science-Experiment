```python
import pandas as pd
data = {
    "Name":["Amit","Riya","Vishal","Yash","Priya"],
    "Age":[20,21,20,22,21],
    "Department":["CSE","IT","CSE","ECE","IT"],
    "Marks":[85,90,78,88,76]
}
df=pd.DataFrame(data)
print(df)
```

```python
print(df.head())
print(df.head())
print("Shape:" , df.shape)
print(df.info())
print(df.describe)
```

```python
df.to_csv("Students.csv",index=False)
```
