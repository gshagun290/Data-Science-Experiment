```python
import pandas as pd

df = pd.read_csv("student_exam_performance.csv")

df.columns = df.columns.str.strip().str.lower()

print("Column Names:")
print(df.columns.tolist())

for column in df.columns:
    if 'math' in column:
        print("\nAverage Math Score:")
        print(df[column].mean())

    if 'reading' in column:
        print("\nAverage Reading Score:")
        print(df[column].mean())

    if 'writing' in column:
        print("\nAverage Writing Score:")
        print(df[column].mean())
```

```python
filtered_df = df[df['previous score'] > 80]

print(filtered_df)

filtered_df.to_csv('/content/students_score_above_80.csv', index=False)

print("Filtered data saved successfully")
```
